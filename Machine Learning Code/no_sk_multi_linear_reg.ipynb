<h1> using normal equation on train test split

```python
def multi_lr(x: pd.DataFrame, y: pd.Series):

    x_train, x_test, y_train, y_test = tts(
        x, y, test_size=0.2, random_state=42
    )

    # Ensure y_train is column vector
    y_train = np.asarray(y_train).reshape(-1, 1)

    # Convert X to NumPy
    x_train = np.asarray(x_train)
    x_test = np.asarray(x_test)

    # Add bias column
    m = x_train.shape[0]
    x_b = np.hstack([np.ones((m, 1)), x_train])

    # Normal Equation
    x_t = x_b.T
    beta = np.linalg.inv(x_t @ x_b) @ (x_t @ y_train)

    print("Intercept:", beta[0])
    print("TV coefficient:", beta[1])
    print("Radio coefficient:", beta[2])
    print("Newspaper coefficient:", beta[3])

    # Prediction
    m_test = x_test.shape[0]
    x_test_b = np.hstack([np.ones((m_test, 1)), x_test])
    y_pred = x_test_b @ beta

    return beta, y_pred, x_test, y_test
```

```python
lr = multi_lr(x,y)
```

```python
from sklearn.metrics import r2_score

# Unpack x_test, y_pred, and y_test from the lr tuple
_, y_pred, x_test, y_test = lr

r2_score(y_test, y_pred)
```

<h1> using Normal equation full dataset training

```python
def multi_lr(x: pd.Series, y: pd.Series):

  # reshape y to n*1
  y = y.reshape(-1,1)
  y.shape

  # adding bias column into x
  m = x.shape[0]
  x_b = np.hstack([np.ones((m,1)),x])

  # transposing of x_b
  x_t = x_b.T

  # x_t * x_b
  xtx = x_t @ x_b

  # inverse of xtx
  xtx_inv = np.linalg.inv(xtx)

  # x_t * y
  xt_y = x_t @ y

  # final beta computation
  beta = xtx_inv @ xt_y

  print("Intercept:", beta[0])
  print("TV coefficient:", beta[1])
  print("Radio coefficient:", beta[2])
  print("Newspaper coefficient:", beta[3])

  y_pred = x_b @ beta

  return beta, y_pred
```

<h1> using class

```python
x_train, x_test, y_train, y_test = tts(x,y,test_size=0.2,random_state=42)
```

```python
y_train.shape
```

```python
class mlr:

  def __init__(self):
    self.coeff_ = None
    self.Intercept = None

  def train(self, x_train, y_train):
    x_train = np.insert(x_train, 0, 1, axis=1)

    beta = np.linalg.inv(np.dot(x_train.T, x_train)).dot(x_train.T).dot(y_train)

    self.coeff_ = beta[1:]
    self.Intercept = beta[0]

  def predict(self,x_test):
    y_pred = np.dot(x_test,self.coeff_) + self.Intercept
    return y_pred
```

```python
lr = mlr()
```

```python
lr.train(x_train,y_train)
```

```python
prediction = lr.predict(x_test)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test,prediction)
```
