```python
from math import inf
import random
x_scaled = (x - x.mean()) / x.std()
x_train,x_test,y_train,y_test = tts(x_scaled,y, test_size=0.2, random_state=42)



class MBGDRegressor:

  def __init__(self,learning_rate=0.01,epochs=100, tolerance = 1e-17, patience = 15,batch_size=10):
    self.learning_rate = learning_rate
    self.epochs = epochs
    self.coef_ = None
    self.intercept_ = None
    self.tolerance = tolerance
    self.patience = patience
    self.batch_size = batch_size

  def fit(self,x_train,y_train):

    # initializing random coef & intercept
    self.coef_ = np.zeros(x_train.shape[1])
    self.intercept_ = 0

    previous_loss = float(inf)
    patience_counter = 0

    print(self.coef_,self.intercept_)

    for i in range(self.epochs):
      for j in range(int(x_train.shape[0]/self.batch_size)):

        idx = np.random.choice(x_train.shape[0],self.batch_size, replace=False)

        y_hat = np.dot(x_train.iloc[idx], self.coef_) + self.intercept_

        # Loss (MSE)
        loss = np.mean((y_train.iloc[idx] - y_hat) ** 2)

        intercept_der = -2 * np.mean(y_train.iloc[idx] - y_hat)
        self.intercept_ -= self.learning_rate * intercept_der

        coef_der = -2 * np.dot((y_train.iloc[idx] - y_hat), x_train.iloc[idx])
        self.coef_ -= self.learning_rate * coef_der

        # print control
      if i < 200 or i >= self.epochs - 200:
            print(f"epoch={i}, coef={self.coef_}, intercept={self.intercept_}")


        # early stopping
      if abs(previous_loss - loss) < self.tolerance:
        patience_counter +=1
        if patience_counter >= self.patience:
          print(f"Early stopping at epochs {i}")
          break
      else:
        patience_counter = 0
        previous_loss = loss

  def predict(self,x_test):
    return np.dot(x_test,self.coef_) + self.intercept_
```

```python
from math import inf
import random
x_scaled = (x - x.mean()) / x.std()
x_train,x_test,y_train,y_test = tts(x_scaled,y, test_size=0.2, random_state=42)



class MBGDRegressor:

  def __init__(self,learning_rate=0.01,epochs=100, tolerance = 1e-17, patience = 15,batch_size=10):
    self.learning_rate = learning_rate
    self.epochs = epochs
    self.coef_ = None
    self.intercept_ = None
    self.tolerance = tolerance
    self.patience = patience
    self.batch_size = batch_size
    self.history = []

  def fit(self,x_train,y_train):

    # initializing random coef & intercept
    self.coef_ = np.zeros(x_train.shape[1])
    self.intercept_ = 0

    previous_loss = float(inf)
    patience_counter = 0

    print(self.coef_,self.intercept_)

    for i in range(self.epochs):
      for j in range(int(x_train.shape[0]/self.batch_size)):

        params = np.hstack([self.coef_, self.intercept_])
        self.history.append(params)


        idx = np.random.choice(x_train.shape[0],self.batch_size, replace=False)

        y_hat = np.dot(x_train.iloc[idx], self.coef_) + self.intercept_

        # Loss (MSE)
        loss = np.mean((y_train.iloc[idx] - y_hat) ** 2)

        intercept_der = -2 * np.mean(y_train.iloc[idx] - y_hat)
        self.intercept_ -= self.learning_rate * intercept_der

        coef_der = -2 * np.dot((y_train.iloc[idx] - y_hat), x_train.iloc[idx])
        self.coef_ -= self.learning_rate * coef_der

        # print control
      if i < 200 or i >= self.epochs - 200:
            print(f"epoch={i}, coef={self.coef_}, intercept={self.intercept_}")


        # early stopping
      if abs(previous_loss - loss) < self.tolerance:
        patience_counter +=1
        if patience_counter >= self.patience:
          print(f"Early stopping at epochs {i}")
          break
      else:
        patience_counter = 0
        previous_loss = loss

  def predict(self,x_test):
    return np.dot(x_test,self.coef_) + self.intercept_
```

```python
history = np.array(gd.history)
history.shape
```

```python
r2 = r2_score(y_test, y_pred)
r2
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
history_2d = pca.fit_transform(history)
```

```python
pc1_vals = np.linspace(
    history_2d[:, 0].min(),
    history_2d[:, 0].max(),
    60
)
pc2_vals = np.linspace(
    history_2d[:, 1].min(),
    history_2d[:, 1].max(),
    60
)

PC1, PC2 = np.meshgrid(pc1_vals, pc2_vals)
```

```python
def inverse_pca(pc1, pc2, pca, mean):
    pc = np.array([pc1, pc2])
    return pca.inverse_transform(pc) + mean
```

```python
def loss_from_pca(pc1, pc2, pca, mean, x, y):
    params = inverse_pca(pc1, pc2, pca, mean)
    w = params[:-1]
    b = params[-1]
    y_hat = np.dot(x, w) + b
    return np.mean((y - y_hat) ** 2)
```

```python
Z = np.zeros_like(PC1)

for i in range(PC1.shape[0]):
    for j in range(PC1.shape[1]):
        Z[i, j] = loss_from_pca(
            PC1[i, j],
            PC2[i, j],
            pca,
            history.mean(axis=0),
            x_train.values,
            y_train.values
        )
```

```python
import plotly.graph_objects as go

surface = go.Surface(
    x=PC1,
    y=PC2,
    z=Z,
    colorscale="Viridis",
    opacity=0.85
)

path = go.Scatter3d(
    x=history_2d[:, 0],
    y=history_2d[:, 1],
    z=[
        loss_from_pca(
            history_2d[i, 0],
            history_2d[i, 1],
            pca,
            history.mean(axis=0),
            x_train.values,
            y_train.values
        )
        for i in range(len(history_2d))
    ],
    mode="lines+markers",
    marker=dict(size=4, color="red"),
    name="MBGD Path"
)

fig = go.Figure(data=[surface, path])

fig.update_layout(
    title="PCA-2D Loss Surface with Mini-Batch GD Trajectory",
    scene=dict(
        xaxis_title="PC1",
        yaxis_title="PC2",
        zaxis_title="MSE Loss"
    )
)

fig.show()
```
