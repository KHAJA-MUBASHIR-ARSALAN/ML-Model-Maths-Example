```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('placement.csv')
```

```python
df
```

```python
x = df['cgpa']
y = df['package']
```

```python
x
```

```python
# st-01 calculate mean
def linear(x: pd.DataFrame, y:pd.DataFrame):
  x_mean = np.mean(x)
  y_mean = np.mean(y)

  # st-02 calculate deviateion

  # numerator

  numerator = np.sum((x-x_mean) * (y-y_mean))

  # denominator

  denominator = np.sum((x-x_mean)**2)

  # st-03 calculating slope M

  m = numerator / denominator

  # st-04 calculating intercept b

  b = y_mean - m * x_mean

  print(f"slope is:{m}")
  print(f"intercept is:{b}")
  y_pred = m*x + b

  display(y_pred)
  values = {
      "slope":m,
      "intercept":b,
      "y_pred":y_pred,
      "y_mean": y_mean
  }
  return values
```

```python
ln = linear(x,y)
```

```python
import matplotlib.pyplot as plt

def plot(x: pd.DataFrame, y: pd.DataFrame):
  plt.scatter(x, y, label="Actual Data")
  plt.plot(x, ln['y_pred'], color="red", label="Regression Line")

  plt.xlabel("CGPA")
  plt.ylabel("Package")
  plt.title("CGPA vs Package (Linear Regression)")
  plt.legend()
  plt.show()

plot(x,y)
```

```python
ss_res = np.sum((y - ln['y_pred']) ** 2)
ss_tot = np.sum((y - ln['y_mean']) ** 2)

r2 = 1 - (ss_res / ss_tot)
print("R² Score:", r2)
```
