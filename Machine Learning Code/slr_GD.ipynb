```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

<h1> sklearn linear reg

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error


x_centered = x - x.mean()


lr = LinearRegression()


lr.fit(x_centered,y)

print("Coefficient is:", lr.coef_)

print("Intercept is:", lr.intercept_)

plt.scatter(x_centered,y)
plt.plot(x_centered,lr.predict(x_centered),color='red')

y_pred_lr = lr.predict(x_centered)
print("sklearn MSE:", mean_squared_error(y, y_pred_lr))
```

```python
m_original = lr.coef_[0][0] / x.std()
b_original = lr.intercept_ - m_original * x.mean()

print(m_original, b_original)
```

```python
class GD:
    def __init__(self, learning_rate, epochs):
        self.lr = learning_rate
        self.epochs = epochs
        self.m = 0.0
        self.b = 0.0
        self.losses = []

    def fit(self, x, y):
        x = x.ravel()
        y = y.ravel()
        n = len(y)

        # reset parameters on every fit
        self.m = 0.0
        self.b = 0.0
        self.losses = []

        for i in range(self.epochs):
            y_pred = self.m * x + self.b

            error = y - y_pred
            loss = np.mean(error ** 2)
            self.losses.append(loss)

            dm = (-2 / n) * np.sum(x * error)
            db = (-2 / n) * np.sum(error)

            self.m -= self.lr * dm
            self.b -= self.lr * db

            # safety break
            if not np.isfinite(loss):
                print("Diverged at epoch", i)
                break

    def predict(self, x):
        return self.m * x + self.b




x_mean = x.mean()
x_std = x.std()

x_norm = (x - x_mean) / x_std

gd = GD(learning_rate=0.05, epochs=3000)
gd.fit(x_norm, y)


m = gd.m / x_std
b = gd.b - m * x_mean

y_pred_gd = m * x + b
print("GD MSE:", mean_squared_error(y, y_pred_gd))

plt.scatter(x_norm, y)
plt.plot(x_norm, gd.predict(x_norm), color='red')
plt.show()



```
